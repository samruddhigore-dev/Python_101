```python
## Q.10 Display grades (A/B/C/D/Fail) based on percentage.
percentage = float(input("Enter percentage: "))

if percentage >= 90:
    print("Grade A")
elif percentage >= 75:
    print("Grade B")
elif percentage >= 60:
    print("Grade C")
elif percentage >= 40:
    print("Grade D")
else:
    print("Fail")
```

```python
## Q11. Build a calculator using if-elif for +, -, *, /
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result =", num1 + num2)
elif operator == "-":
    print("Result =", num1 - num2)
elif operator == "*":
    print("Result =", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result =", num1 / num2)
    else:
        print("Division by zero is not allowed.")
else:
    print("Invalid operator.")
```

```python
## Q.12 Calculate electricity bill using slab rates
units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = (100 * 2) + (units - 100) * 3
elif units <= 300:
    bill = (100 * 2) + (100 * 3) + (units - 200) * 5
else:
    bill = (100 * 2) + (100 * 3) + (100 * 5) + (units - 300) * 7

print("Electricity Bill = ₹", bill)
```
